# Week 3 — Data Visualization and Reporting
Gapminder country-level development dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../data/gapminder_cleaned.csv')
df.head()
```

## Visualization strategy
The visualizations focus on trends, comparisons, relationships, and patterns. The main chart types are bar charts, line charts, scatter plots, and heatmaps.

## 1. Bar chart — continental comparison

```python
d2007=df[df['year']==2007].groupby('continent')['lifeexp'].mean().sort_values(ascending=False)
plt.figure(figsize=(10,6)); plt.bar(d2007.index,d2007.values)
plt.title('Average Life Expectancy by Continent — 2007'); plt.xlabel('Continent'); plt.ylabel('Average Life Expectancy (years)')
plt.tight_layout(); plt.show()
```

**Interpretation:** This comparison highlights differences in average life expectancy between continents in 2007.

## 2. Line chart — time trend

```python
trend=df.groupby('year')['lifeexp'].mean()
plt.figure(figsize=(10,6)); plt.plot(trend.index,trend.values,marker='o')
plt.title('Global Average Life Expectancy Over Time'); plt.xlabel('Year'); plt.ylabel('Average Life Expectancy (years)'); plt.grid(alpha=.25)
plt.tight_layout(); plt.show()
```

**Interpretation:** The line chart shows the long-term direction of average life expectancy and makes changes over time easier to communicate.

## 3. Scatter plot — relationship

```python
s=df[df['year']==2007]
plt.figure(figsize=(10,6))
for c,g in s.groupby('continent'): plt.scatter(g['gdp_per_capita'],g['lifeexp'],label=c,alpha=.7)
plt.xscale('log'); plt.title('GDP per Capita vs Life Expectancy — 2007'); plt.xlabel('GDP per Capita (log scale)'); plt.ylabel('Life Expectancy (years)'); plt.legend(title='Continent')
plt.tight_layout(); plt.show()
```

**Interpretation:** The scatter plot is used to examine whether countries with different GDP-per-capita levels tend to have different life expectancy levels. This is an association, not proof of causation.

## 4. Heatmap — pattern across time and geography

```python
pivot=df.pivot_table(index='continent',columns='year',values='lifeexp',aggfunc='mean')
plt.figure(figsize=(12,5)); plt.imshow(pivot.values,aspect='auto'); plt.colorbar(label='Average Life Expectancy')
plt.xticks(range(len(pivot.columns)),pivot.columns,rotation=45); plt.yticks(range(len(pivot.index)),pivot.index)
plt.title('Average Life Expectancy by Continent and Year'); plt.xlabel('Year'); plt.ylabel('Continent'); plt.tight_layout(); plt.show()
```

**Interpretation:** The heatmap gives a compact view of how life expectancy differs between continents and changes across the available years.

## 5. Additional visualization — top populations

```python
top=df[df['year']==2007].nlargest(10,'pop').sort_values('pop')
plt.figure(figsize=(10,6)); plt.barh(top['country'],top['pop']/1e6)
plt.title('Top 10 Countries by Population — 2007'); plt.xlabel('Population (millions)'); plt.ylabel('Country'); plt.tight_layout(); plt.show()
```

## Key communication principles
Use descriptive titles, clearly labeled axes, readable legends, appropriate scales, and a restrained number of charts. Each chart should answer a specific analytical question rather than simply decorate the report.
